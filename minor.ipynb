Section 2. CONDITIONALS - IF/ELIF/ELSE

2.1. if/else

```python
age = 20

if age >= 18:
  print("Adult")
else:
  print("Minor")
```

2.2. if/elif/else -- more than two outcomes

```python
marks = 72

if marks > 90:
  grade = "A"
elif marks >= 75:
  grade = "B"
elif marks >= 60:
  grade = "C"
else:
  grade = "F"

print(f"Grade:{grade}")
```

```python
print("Grade:" ,grade)
```

2.3. COMBINING CONDITIONS WITH AND / OR

```python
age = 20
has_id = True

if age >= 18 and has_id:
  print("Entry allowed")
else:
  print("Entry denied")
```

Section 3 . Loops - for,while,break,continue

3.1. for loop over a list

```python
fruits = ["Apple", "Banana", "Cherry"]

for x in fruits:
  print(x)
```

3.2. range() and enumerated()

```python
for i in range(5):
  print(i)

print("--")

fruits = ["Apple", "Banana", "Cherry"]
for index,fruit in enumerate(fruits):
  print(index,fruit)
```

3.3. while loop

```python
count = 1

while count <= 10:
  print(count)
  count += 1
```

3.4. break and continue

```python
for num in range(1,10):
  if num == 5:
     break
  print(num)

print("--")

for num in range(1,10):
    if num % 2 == 0:
       continue
    print(num)
```

Practice Time - Sum of even numbers


```python
numbers = [4,7,12,9,18,3,22]
total = 0

for n in numbers:
  if n % 2 != 0:
    continue
  total += n

print(f"Sum of evens: {total}")
```

Section 4. Function

4.1. Defining and calling a function

```python
def greet(name):
  print(f"Hello,{name}!")

greet("Olaf")
greet("Yeti")
```

4.2. return values

```python
def add(a,b):
  return a + b

result = add(5, 3)
print(result)
```

4.3. Default arguments

```python
def greet(name, greeting="Hello"):
  print(f"{greeting}, {name}!")

greet("Olaf")
greet("Yeti", "Namaste")
```

4.4. Keyword argument

```python
def student_info(name, age, city):
  print(f"{name}, {age}, {city}")

student_info(name="Olaf", city="Delhi", age=21)
```

4.5. Variables scope - local vs global

```python
x = 10

def show():
  x = 5
  print("Inside function:", x)

show()
print("Outside function:", x)
```
