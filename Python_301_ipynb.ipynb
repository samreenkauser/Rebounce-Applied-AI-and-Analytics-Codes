```python
import pandas as pd
import numpy as np

url = "https://huggingface.co/datasets/aarav912/online-retail/resolve/main/online_retail.csv"
df = pd.read_csv(url)
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df = df.dropna(subset=["CustomerID"])
```

```python
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
```

```python
df["TotalAmount"] = df["Quantity"] * df["UnitPrice"]
```

```python
df["Sales"] = df["Quantity"] * df["UnitPrice"]
```

```python
df = df.drop_duplicates()
```

```python
df = df[(df["Quantity"] > 0) & (df["UnitPrice"] > 0)]
```

```python
sales_above_500 = df[df["Sales"] > 500]
print("Number of transactions above 500:", len(sales_above_500))
```

```python
top_10_transactions = df.sort_values("Sales", ascending=False).head(10)
print(top_10_transactions)
```

```python
germany = df[df["Country"] == "Germany"]
print("Number of transactions from Germany:", len(germany))
```

```python
print("Mean:", np.mean(df["Sales"]))
print("Median:",
      np.median(df["Sales"]))
print("Minimum:",
      np.min(df["Sales"]))
print("Maximum:",
      np.max(df["Sales"]))
print("Standard Deviation:",
      np.std(df["Sales"]))
```

### Numpy Analysis
  The mean sales value is higher than the median sales value. This indicates that the sales distribution is positively skewed, because a small number of very large transactions iincrease the mean. The maximum sales value is much higher than the median, showing the presence of unusually large transactions.

```python
large_quantity = df[df["Quantity"] > 1000]
print(large_quantity)
```

```python
summary = df.groupby("Country").agg(Number_of_Transactions=("Sales", "count"),
                                    Total_Sales=("Sales", "sum"),
                                    Average_Sales=("Sales", "mean")).reset_index()
print(summary)
```

```python
top_5_countries = summary.sort_values("Total_Sales", ascending=False).head(5)
print(top_5_countries)
```

```python
highest_average = summary.sort_values("Average_Sales", ascending=False).head(1)
print(highest_average)
```

```python
top_country = summary.sort_values("Total_Sales", ascending=False).iloc[0]

print("Top-selling country:", top_country["Country"])
print("Number of transactions:", top_country["Number_of_Transactions"])
```

```python
region_lookup = {
    "United Kingdom": "Europe",
    "France": "Europe",
    "Germany": "Europe",
    "Spain": "Europe",
    "Netherlands": "Europe",
    "Belgium": "Europe",
    "Switzerland": "Europe",
    "Portugal": "Europe",
    "Australia": "Oceania",
    "Japan": "Asia",
    "USA": "Americas"}

regions = pd.DataFrame(list(region_lookup.items()), columns=["Country", "Region"])

print(regions)
```

```python
summary = pd.merge(summary, regions, on="Country", how="left")

summary["Region"] = summary["Region"].fillna("other")

print(summary)
```

```python
region_summary = summary.groupby("Region").agg(
    Number_of_Transactions=("Number_of_Transactions", "sum"),
    Total_Sales=("Total_Sales", "sum"),
    Average_Sales=("Average_Sales", "mean")
).reset_index()

print(region_summary)
```

```python
region_sales = summary.groupby("Region")["Total_Sales"].sum().sort_values(ascending=False)

print(region_sales)
```

```python
print("Best-performing region:", region_sales.idxmax())
print("Total sales:", region_sales.max())
```

4. The UK generated the highest total sales,becoz it had a very large number of transactions,while the Netherlanda had a much higher average transaction value.
5. Missing CustomerID values were removed. Duplicate records & transactions with non-positive Quantity were also removed.


Final Insight -

  The UK generated the highest total sales, while Europe was the best-performing region overall. The Netherlands had the highest average transaction value despite having much lower total sales than the UK. The mean sales valaue was higher than the median, indicatiing that a small number of very large transactions affected the overall distribution.

## Data Analysis Summary

   - The dataset contains online retail transaction data.
   - Missing Customer ID values were removed.
   - InvoiceDate was converted to datetime format.
   - Duplicate records were removed.
   - Transactions with Quantity and UnitPrice greater than 0 were retained.
   - Total sales were calculated as Quantity * UnitPrice.
