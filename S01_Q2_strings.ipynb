Exercise 10 - Clean up a messy name . hard

```python
messy = "   aSHa   verma   "

# Step 1: remove the spaces at both ends.
# Step 2: fix the capitalisation.
# Step 3: squeeze the double space in the midddle down to one.

# Combine steps: strip, convert to lowercase, split by any whitespace, join with single space, then title case.
clean = " ".join(messy.strip().lower().split()).title()

print (f"[{clean}]")
```

Exercise 11 - Split a record . hard

```python
row = "Asha, Delhi, 88"

# split() cuts a string into a list , wherever it finds the seperator.
parts = row.split(",")
print(parts)

name = parts[0]
city = parts[1]
marks = int(parts[2])

print(f"{name}, from {city} scored {marks}")
print(marks + 2)
```
