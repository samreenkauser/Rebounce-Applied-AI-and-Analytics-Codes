**Exercise 10 = Filter states by literacy (Boolean Masking)**

```python
import numpy as np
import pandas as pd

state_data = pd.read_csv("state_indicators.csv")

# Correcting the comparison for 'literacy_rate'
# This creates a boolean Series (mask) where True means literacy_rate >= 85
literacy_rate_mask = state_data['literacy_rate'] >= 85
print(literacy_rate_mask)is True
rates = state_data["literacy_rate"].to_numpy()
print(rates[literacy_rate_mask])
# Correcting the second print statement to count how many states meet the criteria
# .sum() on a boolean Series counts the number of True values
print(literacy_rate_mask.sum())
```
