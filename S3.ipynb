# **DICTIONARY

Lists are for sequences.But what if you want to store details about a single entity?
A Dictionary uses "Keys"(labels) and "Values". It's like a row in your Excel sheet.

```python
district_data = {
    "Name": "Pune",
    "States": "Maharastra",
    "Literacy_Rate": 86.1,
    "Urban": True
}
```

```python
print(district_data)
```

```python
print(district_data["Name"])
```

```python
print(district_data["Name"],"-",district_data["Literacy_Rate"])
```

```python
district_data.keys()
```

```python
district_data.values()
```

```python
district_data["collector"] = "Dr. Haathi"
```

```python
district_data
```

```python
district_data["Literacy_Rate"] = 89.8
district_data
```

```python
del district_data["collector"]
```
