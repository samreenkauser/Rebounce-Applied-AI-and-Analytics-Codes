EXERCISE 2 - SLICING

```python
name = "Data Science"

# Correcting string slicing with integer indices
# To get 'Data', you'd slice from index 0 up to (but not including) 4
print(name[0:4])
# To get 'Science', you'd slice from index 5 to the end
print(name[5:])
# You can also use negative indices, e.g., name[-7:] for 'Science'
```

Exercise 3 - Stride

```python
s = "Micheal Jackson"

# Corrected first line: example of slicing with start, stop, and step
# This will print 'Mce sc'
print(s[0:15:2])

# Corrected second line: example of slicing in reverse
# This will print 'noskcaJ laehciM' (reversed string)
print(s[::-1])

# To get a slice from index 12 to 0 in reverse with a step of -1:
# This will print 'noskcaJ laehci' (from index 12 up to, but not including, index 0, in reverse)
print(s[12:0:-1])
```

Exercise 5 - Find it

```python
s = "Micheal Jackson"

# To find the starting index of a substring 'el'
# The find() method returns the lowest index where the substring is found, or -1 if not found.
print(s.find("el"))

# To find the starting index of a substring 'Jackson'
print(s.find("Jackson"))

# To check if a substring 'o' exists in the string or to find its position
print(s.find("o"))

# If you meant to replace parts of the string, you would use replace()
# For example, to replace 'Micheal' with 'Michael'
print(s.replace("Micheal", "Michael"))

# Another example: replacing 'o' with 'O'
print(s.replace("o", "O"))
```

Exercise 11 - Split a record

```python
row = "Asha,Delhi,88"
parts = row.split(',') # Corrected split method
print(parts)

name = parts[0] # Accessing by integer index
city = parts[1] # Accessing by integer index
marks = int(parts[2]) # Converting to integer

print(f"{name} from {city} scored {marks}") # Added space
print(marks + 2)
```

Exercise 12 - Is it a palindrome

```python
word = "Malyalam"

# Corrected: use [::-1] to reverse the string
print(word.upper() == word[::-1].upper())

word = "Python"
# Corrected: to check if 'Python' is a palindrome in lowercase
print(word.lower() == word[::-1].lower())
```
