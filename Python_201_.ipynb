# Minor project 2 - Personal Expenses Tracker

#Track A - Personal Expense Tracker
This program helps users record their daily expenses. It displays expenses and calculates total spending and category-wise spending. It saves expenses to a file and loads them again when the program starts.

### How to run Run all cells, then call `main()` to start program.


```python
expenses = []
print("Expenses tracker started")
```

1. Dictionary = one expense

```python
expense = {"category": "food", "amount": 250, "description": "Lunch at the office canteen"}
expenses.append(expense)

print(expense)
```

2. List = all expenses

```python
print(expenses)
```

```python
total = 0
for expense in expenses:
  total += expense["amount"]
  print("Total spent:", total)
```

```python
category_totals = {}
for expense in expenses:
  category = expense["category"]
  amount = expense["amount"]
  if category in category_totals:
    category_totals[category] += amount
  else:
    category_totals[category] = amount

print("Category Totals:")
for category, total_amount in category_totals.items():
  print(f"{category}: {total_amount}")
```

```python
def add_expenses():
  category = input("Enter expense category: ").strip()

  if not category:
    print("Category cannot be empty.")
    return

  try:
    amount = float(input("Enter expense amount: "))
    if amount < 0:
      print("Amount cannot be negative.")
      return
  except ValueError:
    print("Invalid amount. Please enter a number.")
    return

  description = input("Enter expense description: ").strip()

  if not description:
    print("Description cannot be empty.")
    return

  expense = { "category": category, "amount": amount, "description": description}

  expenses.append(expense)
  print("Expense added.")
```

```python
def view_expenses():
  if not expenses:
    print("No expenses recorded.")
  else:
    print("Expenses:")
    for expense in expenses:
      print(f"  Category: {expense['category']}")
      print(f"  Amount: {expense['amount']}")
      print(f"  Description: {expense['description']}")
      print()
```

```python
def save_expenses():
  with open("expenses.txt", "w") as file:
    for expense in expenses:
      file.write(f"{expense['category']}, {expense['amount']}, {expense['description']}\n")
      print("Expenses saved.")
```

```python
def load_expenses():
  expenses.clear()
  try:
    with open("expenses.txt", "r") as file:
      for line in file:
        category, amount, description = line.strip().split(",")
        expense = {"category": category, "amount": amount, "description": description}
        expenses.append(expense)
    print("Expenses loaded.") # Move print outside the loop to avoid printing for each line
  except FileNotFoundError:
    print("No saved expenses found.")
```

```python
def view_category_totals():
  category_totals = {}
  for expense in expenses:
    category = expense["category"]
    # Ensure amount is treated as a number for calculation
    amount = float(expense["amount"])
    if category in category_totals:
      category_totals[category] += amount
    else:
      category_totals[category] = amount

  print("\nCategory Totals:")
  if not category_totals:
    print("  No category totals to display.")
  else:
    for category, total_amount in category_totals.items():
      print(f"  {category}: {total_amount:.2f}")

def main():
  load_expenses()

  while True:
    print("\n=== Expense Tracker ===")
    print("1. Add Expense")
    print("2. View all expenses")
    print("3. View total by category")
    print("4. View total spent")
    print("5. Save and quit")
    choice = input("Enter your choice: ")

    if choice == "1":
      add_expenses()
    elif choice == "2":
      view_expenses()
    elif choice == "3":
      view_category_totals() # Call the new function here
    elif choice == "4":
      total = 0
      for expense in expenses:
        # Ensure amount is treated as a number for calculation
        total += float(expense["amount"])
      print(f"Total spent: Rs {total:.2f}")
    elif choice == "5":
      save_expenses()
      break
    else:
      print("Invalid choice. Please enter 1-5.")
```

```python
main()
```
